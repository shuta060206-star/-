```python
# Q18

    loaded = json.load(f)
print(loaded["osaka"]["temp"])

# Q19
total = 0
with open("squares.csv", "r") as f:
    f.readline()                      # ヘッダー行を読み飛ばす
    for line in f:
        parts = line.strip().split(",")
        total = total + int(parts[1])
print(total)

# Q20
with open("seiseki.csv", "w") as f:
    f.write("鈴木,82,91,76\n田中,90,68,85\n佐藤,77,95,88\n")

best_name = None
best_mean = -1                        # どんな平均にも負ける初期値から始める
with open("seiseki.csv", "r") as f:
    for line in f:
        parts = line.strip().split(",")
        name = parts[0]
        scores = [int(s) for s in parts[1:]]   # リスト内包表記(短縮技。forとappendでも可)
        mean = sum(scores) / len(scores)
        print(f"{name}: {mean:.1f}")
        if mean > best_mean:                   # 暫定チャンピオン方式
            best_mean = mean
            best_name = name
print("トップ:", best_name)
```
