# Q15
x = np.linspace(0, 0.3, 31)
T = np.zeros(31)
T[0] = T[-1] = 1.0
for step in range(1, 51):
    T_new = T.copy()
    T_new[1:-1] = T[1:-1] + 0.4 * (T[:-2] - 2*T[1:-1] + T[2:])
    T = T_new
    if step % 10 == 0:                       # 10回ごと(余りが0のとき)
        plt.plot(x, T, label=f"step {step}")
plt.xlabel("x [m]")
plt.ylabel("T")
plt.legend()
plt.grid(True)
plt.show()
