```python
def check_password(pw: str) -> str:
    """パスワード強度を判定して説明文を返す"""
    if len(pw) < 8:
        return "弱い: 8文字以上にしてください"

    has_digit = False
    has_upper = False
    for ch in pw:                    # フラグ変数パターン: 見つけたら立てる
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_upper = True

    if not has_digit:
        return "普通: 数字を混ぜると強くなります"
    if has_upper:
        return "強い"
    return "普通: 大文字を混ぜると強くなります"

for pw in ["abc", "abcdefgh", "abcdefg1", "Abcdefg1"]:
    print(f"{pw}: {check_password(pw)}")
```

```python
class Inventory:
    """商品名→個数の在庫を管理する"""

    def __init__(self):
        self.stock = {}

    def add(self, name: str, count: int):
        if count <= 0:
            raise ValueError(f"追加数は正の値が必要です: {count}")
        self.stock[name] = self.stock.get(name, 0) + count

    def remove(self, name: str, count: int):
        if name not in self.stock:
            raise ValueError(f"未登録: {name}")
        if self.stock[name] < count:
            raise ValueError(f"在庫不足: {name} は {self.stock[name]}個しかありません")
        self.stock[name] = self.stock[name] - count

    def report(self):
        for name, count in self.stock.items():
            print(f"{name}: {count}個")


inv = Inventory()
inv.add("りんご", 12)
inv.add("みかん", 30)
inv.remove("りんご", 5)
inv.report()

for name, count in [("みかん", 100), ("バナナ", 1)]:
    try:
        inv.remove(name, count)
    except ValueError as e:
        print("エラー:", e)
```

```python
import numpy as np
import matplotlib.pyplot as plt

materials = {
    "concrete": {"lam": 1.6,  "rho": 2200, "c": 880},
    "brick":    {"lam": 0.62, "rho": 1650, "c": 840},
    "wood":     {"lam": 0.12, "rho": 500,  "c": 1300},
}

def simulate(a: float, hours: float, n: int = 30, L: float = 0.3) -> np.ndarray:
    """初期0℃・両端1℃の壁の hours 時間後の温度分布を返す"""
    dx = L / n
    dt = dx ** 2 / (2 * a) * 0.4        # 安定条件に対して余裕を持ったdtを自動選択
    T = np.zeros(n + 1)
    T[0] = T[-1] = 1.0
    r = a * dt / dx ** 2
    steps = int(hours * 3600 / dt)
    for _ in range(steps):
        T_new = T.copy()
        T_new[1:-1] = T[1:-1] + r * (T[:-2] - 2*T[1:-1] + T[2:])
        T = T_new
    return T

x = np.linspace(0, 0.3, 31)
for name, props in materials.items():
    a = props["lam"] / (props["rho"] * props["c"])
    T = simulate(a, hours=12)
    plt.plot(x * 1000, T, label=name)

plt.xlabel("x [mm]")
plt.ylabel("T [C]")
plt.title("Which material insulates best? (after 12 h)")
plt.legend()
plt.grid(True)
plt.show()
# 考察: 中央の温度が最も低い材料が「熱を通しにくい=断熱に優秀」。
# 木材は熱拡散率が最小(λが小さくc·ρあたりの熱容量も効く)ため、中央が最も温まりにくい。
```
