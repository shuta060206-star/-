## 3-1. 比較演算子と bool

まず「条件」を書くための道具。比較の結果は `True`(真)か `False`(偽)という特別な値になります。この2値を **bool(ブール)値** といいます。

````python
x = 10
print(x > 5)     # True
print(x < 5)     # False
print(x == 10)   # 等しいか。「=」1個は代入なので、比較は「==」2個!
print(x != 3)    # 等しくないか → True
print(x >= 10)   # 以上 → True
````

⚠️ **`=`(代入)と `==`(比較)の混同は、全世界の初心者が必ず一度はやるミス**です。「イコール2個で質問、1個で命令」と覚えましょう。

## 3-2. if / elif / else — 条件分岐

形はforループと同じ「コロン + 字下げ」です。

````python
T = 25   # 気温 [℃]

if T >= 30:
    print("暑い!")
elif T >= 15:            # elif = 「そうではなくて、もし〜なら」(いくつでも並べられる)
    print("快適")
else:                    # どれにも当てはまらなかったとき
    print("寒い")
````

条件は `and` / `or` / `not` で組み合わせられます。

````python
T = 20
shitsudo = 65   # 湿度 [%]

if T >= 18 and shitsudo <= 70:
    print("過ごしやすい")

if T < 5 or T > 35:
    print("外出注意")
else:
    print("ふつうに外出できる")
````

---
## 3-3. while ループ — 「条件が成り立つ間」繰り返す

forは「回数が決まっている」繰り返し、whileは「**条件を満たす間ずっと**」の繰り返しです。

````python
zandaka = 1000        # 残高 [円]
kaisu = 0

while zandaka > 0:            # 残高がある間、繰り返す
    zandaka = zandaka - 300   # 300円使う
    kaisu = kaisu + 1
    print(f"{kaisu}回目: 残高 {zandaka} 円")

print("残高がなくなりました")
````

`break` で「ループを途中で強制終了」もできます。実務では「条件を満たすまで探す」処理でよく使います。

````python
for i in range(1, 100):
    if i * i > 50:
        print("2乗が50を超える最小の数は", i)
        break            # 見つけたのでループ終了。以降のiは回らない
````

## 3-4. 関数 — 処理に名前をつけて部品化する

ここ、**実務プログラミングの心臓部**です。同じ処理を何度も書かず、`def` で「部品」として定義して使い回します。

```python
def 関数名(引数1, 引数2):
    処理
    return 結果
```

- **引数(ひきすう)** … 関数に渡す入力
- **return** … 関数が返す出力(戻り値)

````python
def en_menseki(r):
    """半径 r の円の面積を返す"""      # ← docstring: 関数の説明文(実務の作法)
    S = 3.14159 * r ** 2
    return S

# 定義した関数は何度でも呼び出せる
print(en_menseki(5))
print(en_menseki(10))
print(en_menseki(1))
````

引数は複数OK。戻り値を変数に受け取って次の計算に使うのが普通の流れです。

````python
def netsu_kakusanritsu(lam, rho, c):
    """熱拡散率 a = λ/(ρc) を返す [m²/s]"""
    return lam / (rho * c)

a_concrete = netsu_kakusanritsu(1.6, 2200, 880)   # コンクリート
a_glass    = netsu_kakusanritsu(1.0, 2500, 750)   # ガラス
print("コンクリート:", a_concrete)
print("ガラス:", a_glass)
````

👀 レッスン1の練習問題1-Bが「関数」に進化しました。**一度関数にすれば、どんな材料でも1行で計算できる** — これが部品化の威力です。

### 引数のデフォルト値

「省略したらこの値を使う」という指定もできます。実務のライブラリ関数はほぼ全部これを使っています。

````python
def aisatsu(name, greeting="こんにちは"):
    print(f"{greeting}、{name}さん!")

aisatsu("田中")                    # greeting は省略 → デフォルトが使われる
aisatsu("佐藤", greeting="おはよう")   # 名前つきで指定もできる
````

## 💼 実務メモ: なぜ関数に分けるのか

実務でコードを関数に分ける理由は3つ:

1. **再利用** … 同じ処理をコピペしない(コピペは修正漏れバグの温床)
2. **テスト** … 部品ごとに「正しく動くか」を確認できる
3. **読みやすさ** … `netsu_kakusanritsu(1.6, 2200, 880)` は、中身を読まなくても何をするか分かる

「**1つの関数は1つの仕事だけ**」が鉄則。50行を超える関数は分割を疑え、と言われます。
