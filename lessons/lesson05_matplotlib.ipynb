# レッスン5: matplotlib — データを目に見える形に 📊

**matplotlib(マットプロットリブ)** はグラフ描画の標準ライブラリです。数値の羅列を人間が理解できる形にする、報告書・論文・プレゼンの必需品。実務では「計算して終わり」はなく、**必ず可視化して人に伝える**ところまでが仕事です。

---
## 5-1. 最小のグラフ

`import matplotlib.pyplot as plt`(この略称も世界標準)して、`plt.plot(x, y)` → `plt.show()`。これだけ。

````python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 50)    # 0〜10 を50点
y = x ** 2

plt.plot(x, y)
plt.show()
````

---
## 5-2. ラベル・タイトル・凡例 — 「読めるグラフ」にする

軸ラベルのないグラフは実務では通用しません(その数字が何なのか誰にも分からない)。最低限つけるもの:

- `plt.xlabel()` / `plt.ylabel()` … 軸の意味と**単位**
- `plt.title()` … グラフの主張
- `plt.legend()` … 複数の線の見分け(各plotに `label=` を付けておく)
- `plt.grid(True)` … 目盛り線(値が読み取りやすくなる)

※ ラベルは英語で書くのが安全です。日本語はフォントの問題で□□□に化けることがあるため(対策は下の実務メモ)。

````python
x = np.linspace(0, 2 * np.pi, 100)

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")     # plotを重ねると同じ図に複数の線が描ける
plt.xlabel("x [rad]")
plt.ylabel("y")
plt.title("Sine and Cosine")
plt.legend()
plt.grid(True)
plt.show()
````

---
## 5-3. 線の見た目とマーカー

````python
x = np.linspace(0, 0.3, 11)
T = 1 - 4 * (x / 0.3) * (1 - x / 0.3)     # 両端1・中央0のカーブ(温度分布の雰囲気)

plt.plot(x, T, marker="o")                 # marker="o" で点も打つ
plt.plot(x, T * 0.5 + 0.5, linestyle="--", color="red")   # 破線・色指定
plt.xlabel("x [m]")
plt.ylabel("T [C]")
plt.grid(True)
plt.show()
````

よく使う指定: `marker="o"`(丸)`"s"`(四角)、`linestyle="--"`(破線)`":"`(点線)、`color="red"` など。

---
## 5-4. 画像として保存する

レポートに貼るには `plt.savefig()`。**`show()` より前に**呼ぶのがポイント。

````python
x = np.linspace(0, 10, 50)
plt.plot(x, x ** 2)
plt.xlabel("x")
plt.ylabel("y")
plt.savefig("graph.png", dpi=150)     # dpi = 解像度。150〜300がレポート向き
plt.show()
# Colabでは左のフォルダアイコン📁から graph.png をダウンロードできます
````
