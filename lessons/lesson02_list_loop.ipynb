## 2-1. リスト — 値の並びをまとめて持つ

`[ ]` で囲んでカンマで区切ると、複数の値をひとまとめにできます。

````python
T = [0, 0, 0, 0, 0]    # 温度の並び [℃] (5点ぶん)
print(T)
print(len(T))           # len() はリストの要素数(length)
````

### 要素の取り出し — インデックス

リストの各要素には番号(**インデックス**)でアクセスします。

⚠️ **超重要: 番号は 0 から始まります。** 最初の要素が `T[0]`、2番目が `T[1]`。ほぼ全てのプログラミング言語がこの「0始まり」ルールです。最初は戸惑いますが、すぐ慣れます。

````python
T = [10, 20, 30, 40, 50]
print(T[0])     # 最初の要素 → 10
print(T[1])     # 2番目 → 20
print(T[4])     # 5番目(最後) → 50
print(T[-1])    # マイナスは後ろから数える。-1 = 最後 → 50
print(T[-2])    # 後ろから2番目 → 40
````

### 要素の書き換え

`T[番号] = 値` で、その場所だけ書き換えられます。

````python
T = [0, 0, 0, 0, 0]
T[0] = 1      # 最初の要素を 1 に
T[-1] = 1     # 最後の要素を 1 に
print(T)      # [1, 0, 0, 0, 1]
````

👀 いま何をしたか気づきましたか? **「初期温度0℃の壁の、両端だけ1℃にする」** — 課題の初期条件・境界条件そのものです。

### 便利な作り方・伸ばし方

````python
T = [0] * 11        # [0]を11個並べたリストを一発で作る
print(T)

x = []              # 空のリスト
x.append(0.0)       # append() = 末尾に追加
x.append(0.03)
x.append(0.06)
print(x)
````

## 2-2. for ループ — 繰り返しの構文

「同じ処理をn回やる」を書くのが `for` 文です。まず形を見てください:

```python
for i in range(5):
    print(i)
```

読み方: 「`i` を 0, 1, 2, 3, 4 と変えながら、字下げされた行を繰り返せ」

**文法のルール(ここ試験に出ます📝):**
1. `for` の行の最後に **コロン `:`** が必須
2. 繰り返したい行は **字下げ(インデント、半角スペース4つ)** する
3. `range(n)` は 0 から n-1 まで(nは含まない!リストの0始まりと辻褄が合う設計)

字下げが構文の一部になっているのはPythonの特徴で、「どこからどこまでを繰り返すか」を字下げで表します。

````python
for i in range(5):
    print("繰り返し", i, "回目")

print("ループを抜けた(字下げしてない行はループの外)")
````

### ループ × リスト = 最強コンビ

インデックス `i` を回しながらリストの各要素を処理するのが定番パターンです。

````python
L = 0.3         # 壁の厚さ [m]
n = 10          # 分割数
dx = L / n      # 分割幅 [m]

for i in range(n + 1):      # i = 0,1,...,10 (点は分割数+1個ある)
    x = i * dx
    print(f"点{i}: x = {x:.2f} m")
````

最後の行の `f"..."` は **f文字列**という機能で、文字列の中に `{変数}` を埋め込めます。`{x:.2f}` は「小数2桁で表示」。実務コードで毎日使われる書き方です。

### 計算結果をリストに貯める

「ループで計算 → append で貯める」も定番パターン。

````python
x_list = []                  # 空のリストを用意
for i in range(n + 1):
    x_list.append(i * dx)    # 計算しては末尾に追加

print(x_list)
````
