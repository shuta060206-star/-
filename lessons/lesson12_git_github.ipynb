## 12-1. Gitとは — 「セーブポイント付きの共同作業システム」

**Git** = ファイルの変更履歴を記録するツール(自分のPC内で動く)
**GitHub** = Gitの履歴をインターネット上で共有するサービス

ゲームで例えると:

| Git用語 | 意味 | ゲームで言うと |
|---------|------|----------------|
| リポジトリ (repo) | 履歴ごと管理されるプロジェクトのフォルダ | セーブデータ一式 |
| コミット (commit) | 変更をひとまとまりの履歴として記録 | セーブポイントを作る |
| プッシュ (push) | 手元のコミットをGitHubへ送る | クラウドセーブ |
| プル (pull) | GitHubの最新を手元に取り込む | クラウドから同期 |
| ブランチ (branch) | 履歴の分岐。本流を壊さず実験できる | 分岐セーブスロット |
| マージ (merge) | 分岐を本流に合流させる | 分岐の成果を本編に反映 |
| プルリクエスト (PR) | 「この変更を本流に入れてください」という提案+レビューの場 | (GitHub上の文化) |

**なぜ必須なのか:** いつでも過去に戻れる(壊しても平気)/ 誰が・いつ・なぜ変えたか全部残る / 複数人が同じコードを同時に触れる。

## 12-2. 基本コマンド(PC環境で使う)

`docs/pc-setup.md` の環境ができたら、ターミナル(PowerShell)でこの流れを体験します。ここは読むだけでOK、実際の操作は下の練習12-Bで。

```bash
# GitHubのリポジトリを自分のPCに複製する(最初の1回だけ)
git clone https://github.com/shuta060206-star/-.git

# --- 日常の作業サイクル(これの繰り返し) ---
git status                     # 今なにが変更されているか確認(困ったらまずこれ)
git add ondo.py                # コミットに含めるファイルを選ぶ(ステージング)
git commit -m "温度変換モジュールを追加"   # メッセージ付きでセーブ
git push                       # GitHubへ送る
git pull                       # 他の場所での変更を取り込む
```

### コミットメッセージの書き方(実務の作法)

- ❌ `update`, `fix`, `aaa` — 何をしたか分からない履歴は無価値
- ✅ 「**何を・なぜ**」を短く: `安定条件チェックを追加(dt過大による発散を防ぐため)`
- 1コミット = 1つのまとまった変更(「グラフ修正と誤字直しとリファクタ」を混ぜない)

## 12-3. ブランチとプルリクエスト — 実務の開発フロー

実務では本流(`main` ブランチ)に直接コミットせず、こう回します:

```
main ──●──────────────●──→   (常に動く状態を保つ)
        \            /
         ●──●──●──●         ← 作業ブランチ(feature/add-graph など)
              作業     PR+レビュー後にマージ
```

1. 作業ブランチを切る: `git switch -c feature/add-graph`
2. コミットを重ねる
3. GitHubに push して**プルリクエスト(PR)**を出す
4. 同僚が**コードレビュー**(レッスン11の作法はここで効く!)
5. OKなら main に**マージ**

このリポジトリでも実は同じことが起きています: 私(Claude)は `claude/...` という作業ブランチにpushしていて、mainは別にあります。

## 12-4. .gitignore — 履歴に入れないものを決める

計算結果のファイルや一時ファイルまでコミットするとリポジトリが汚れます。`.gitignore` というファイルに書いたパターンはGitが無視します:

```
__pycache__/     # Pythonが自動生成するキャッシュ
*.png            # 生成物のグラフ(コードから再生成できるので)
.DS_Store        # Macが勝手に作るファイル
```

「**ソースコードは入れる、生成物は入れない**」が原則です(コードがあれば生成物は再現できるから)。
