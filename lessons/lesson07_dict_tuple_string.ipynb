## 7-1. 辞書 (dict) — 「名前→値」の対応表

リストは「番号で引く」、辞書は **「名前(キー)で引く」** データ構造です。

````python
user = {"name": "Tanaka", "age": 20, "email": "tanaka@example.com"}

print(user["name"])       # キーで値を取り出す
print(user["age"])

user["age"] = 21          # 書き換え
user["city"] = "Tokyo"    # 新しいキーは追加になる
print(user)
````

### キーの存在チェックとループ

````python
user = {"name": "Tanaka", "age": 21, "city": "Tokyo"}

print("age" in user)        # キーがあるか → True
print("phone" in user)      # → False
print(user.get("phone", "未登録"))   # get() = 無ければデフォルト値。実務頻出!

for key, value in user.items():     # 辞書のループはitems()
    print(key, "→", value)
````

### 辞書のネスト — 実務データはだいたいこの形

Web APIが返すデータ(JSON)や設定ファイルは「リストの中に辞書、辞書の中に辞書」という入れ子構造です。この形に慣れると実務データの9割は読めます。

````python
# 材料データベース: 辞書の中に辞書
materials = {
    "concrete": {"lam": 1.6, "rho": 2200, "c": 880},
    "glass":    {"lam": 1.0, "rho": 2500, "c": 750},
    "wood":     {"lam": 0.12, "rho": 500, "c": 1300},
}

print(materials["concrete"]["lam"])    # 外→内の順にキーで潜る

for name, props in materials.items():
    a = props["lam"] / (props["rho"] * props["c"])
    print(f"{name}: a = {a:.2e} m²/s")
````

## 7-2. タプル (tuple) — 変更できないリスト

`( )` で作る、**後から変更できない**値の並び。「セットで1つの意味」のデータ(座標、RGB色など)や、関数が複数の値を返すときに使います。

````python
point = (3, 5)             # 座標(x, y)
print(point[0])            # 取り出しはリストと同じ

x, y = point               # まとめて2つの変数に展開(アンパック)超頻出!
print(x, y)

def saidai_saisho(values):
    """最大値と最小値を両方返す"""
    return max(values), min(values)     # カンマで並べるとタプルで返る

big, small = saidai_saisho([3, 1, 4, 1, 5, 9, 2])
print("最大:", big, "最小:", small)
````

## 7-3. 文字列の操作 — 実務の隠れ主役

実務プログラミングの体感3割は文字列処理です(ファイル名、ログ、CSV、ユーザー入力...)。

````python
s = "  Hello, Python World  "

print(s.strip())            # 前後の空白を除去
print(s.strip().upper())    # 大文字化 (メソッドは繋げられる=メソッドチェーン)
print(s.strip().lower())    # 小文字化
print(s.replace("Python", "Real"))   # 置換

csv_row = "concrete,1.6,2200,880"
parts = csv_row.split(",")            # 区切り文字で分解 → リスト
print(parts)

words = ["2026", "07", "10"]
print("-".join(words))                # リストを区切り文字で連結 → "2026-07-10"
````

### f文字列の書式指定(レポート・ログの必須技術)

````python
a = 8.264462809917355e-07
price = 1234567

print(f"{a:.3f}")      # 小数3桁 → 0.000 (小さすぎて見えない)
print(f"{a:.3e}")      # 指数表記で3桁 → 8.264e-07 ★科学計算はこれ
print(f"{price:,}")    # 3桁カンマ → 1,234,567 ★業務システムはこれ
print(f"{0.857:.1%}")  # パーセント表示 → 85.7%
print(f"{'left':<10}|{'right':>10}|")   # 幅指定で左寄せ/右寄せ(表の整形)
````
