## 1-1. print() — 画面に表示する

まずは定番、画面に文字を表示することから。下のセルを選んで Shift + Enter!

````python
print("こんにちは、Python!")
````

実行すると、セルのすぐ下に `こんにちは、Python!` と出たはずです。

- `print(...)` は「カッコの中身を画面に表示しろ」という命令です
- 文字を表示したいときは `"` (ダブルクォート)で囲みます

では数字も表示してみましょう。数字は `"` で囲みません。

````python
print(300)
print("壁の厚さは", 300, "mm です")   # カンマで区切ると並べて表示できる
````

`#` から行末までは**コメント**といって、プログラムの動作に関係ないメモ書きです。人間(未来の自分)のための注釈ですね。

## 1-2. 変数 — 値に名前をつける

**変数**は「値に名前をつけた箱」です。`名前 = 値` と書くと、値がその名前で保存され、以後は名前で呼び出せます。

⚠️ 数学と違い、`=` は「等しい」ではなく「**右の値を左の名前に入れる**」という意味です。

````python
L = 300      # 壁の厚さ [mm] という変数を作った
print(L)     # 変数名を書くと、中身の 300 が使われる
print("壁の厚さは", L, "mm です")
````

変数の中身は後から入れ替えられます。

````python
x = 10
print(x)     # 10
x = 99       # 上書き!
print(x)     # 99
x = x + 1    # 「今のxに1を足したものを、あらためてxに入れる」→ 100
print(x)
````

最後の `x = x + 1` は数学だとおかしな式ですが、プログラミングでは超頻出です。「今の値を1増やす」という意味になります。

## 1-3. 計算 — Pythonは電卓としても優秀

| 記号 | 意味 | 例 |
|------|------|-----|
| `+` | 足し算 | `7 + 2` → 9 |
| `-` | 引き算 | `7 - 2` → 5 |
| `*` | 掛け算 | `7 * 2` → 14 |
| `/` | 割り算 | `7 / 2` → 3.5 |
| `**` | べき乗 | `7 ** 2` → 49 |

変数どうしでも計算できます。

````python
a = 7
b = 2
print("足し算:", a + b)
print("引き算:", a - b)
print("掛け算:", a * b)
print("割り算:", a / b)
print("2乗:",   a ** 2)
````

計算結果を変数に保存しておくこともできます(むしろこれが普通の使い方)。

````python
L = 300
L = L / 1000     # mm を m に変換して kabe_m に保存
print("壁の厚さ:", L, "m")
````
